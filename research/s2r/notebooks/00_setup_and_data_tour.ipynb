# 00 — Setup & data tour

Imports the existing S2R package and walks the data folders used for:
- **12-task G1 benchmark**
- **Unitree VLA** eval artifacts
- **ESN** train/val splits

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd() / "_lib"))
from bootstrap import setup
ROOT = setup()
print("ROOT", ROOT)

from s2r.experiments.paths import ensure_experiment_dirs, DATA, BENCHMARK
from s2r.experiments.benchmark import load_tasks
from s2r.core.config import load_config

paths = ensure_experiment_dirs()
cfg = load_config(ROOT / "config" / "default.yaml")
tasks = load_tasks()
print("tasks", len(tasks))
print("data", DATA)
display([{k: str(v)} for k, v in list(paths.items())[:8]])
```

```python
import pandas as pd

df = pd.DataFrame([
    {
        "id": t.id,
        "name": t.name,
        "category": t.category,
        "hf_dataset": t.hf_dataset,
        "instruction": t.instruction,
        "folder": str(t.dir),
    }
    for t in tasks
])
df
```

```python
# Count existing episode logs / metrics placeholders
rows = []
for t in tasks:
    rows.append({
        "task": t.id,
        "episodes": len(list((t.dir / "episodes").glob("*"))),
        "metrics": len(list((t.dir / "metrics").glob("*.json"))),
        "videos": len(list((t.dir / "videos").glob("*"))),
    })
pd.DataFrame(rows)
```

## Folder map

| Path | Purpose |
|---|---|
| `data/benchmark/tasks/XX_*` | Per-task episodes, videos, annotations, metrics |
| `data/benchmark/results` | Aggregated score JSON for leaderboard |
| `data/unitree_vla/` | Unitree UnifoLM-VLA checkpoints / preds / evals |
| `data/esn/` | ESN train/val npz + checkpoints + curves |
| `data/raw/` | Live pipeline JSONL from `data_collector` |
